#### 14. Display a summary of the numerical features again. Identify any columns with identical statistics. If such columns exist, verify whether they are true duplicates and keep only the first occurrence, removing all others. [4pts]

```python
display(df.describe(include=['number']))

numeric_cols = df.select_dtypes(include=['number']).columns
duplicate_cols = set()
```

```python
for i in range(len(numeric_cols)):
    col1 = numeric_cols[i]
    if col1 in duplicate_cols:
        continue
    for j in range(i + 1, len(numeric_cols)):
        col2 = numeric_cols[j]
        if df[col1].equals(df[col2]):
            duplicate_cols.add(col2)

df.drop(columns=list(duplicate_cols), inplace=True)
```

#### 25. Do all cities have unique `cityId`? Find cities wuth multiple IDs if any. [3pts]

#### 27. Replace `NaN` values in the `cityId` column with the corresponding `cityId` for cities that have a unique `cityId`. Report how many missing values where in the `cityId` column before and after the replacement. [10pts]

```python
print(f"Missing before: {df['cityId'].isna().sum()}")
city_to_id = df.groupby('city')['cityId'].first()
df['cityId'] = df['cityId'].fillna(df['city'].map(city_to_id))
print(f"Missing after: {df['cityId'].isna().sum()}")
```
