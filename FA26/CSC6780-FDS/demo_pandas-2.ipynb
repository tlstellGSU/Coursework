# Series

A Series is a column-vector. It is an array of individual entries, each of which has a certain value.

```python
ages = pd.Series([18, 35, 42], name="Age")
ages
```

```python
#This is label-based slicing, but in Pandas, slices are end-exclusive when using integer positions (just like Python lists).
#It takes rows from index 1 up to (not including) index 2.

print(ages[1:2])

#.iloc means integer-location based indexing only!
# iloc[1:2] also takes rows at positions 1 up to (not including) 2.

ages.iloc[1:2]
#ages.loc[1:2]
```

```python
#A plain Python list with mixed types: integers and a string
array = [1, 5, 'hello', 4, -10]

col_vec = pd.Series(array, name="Something")
col_vec

# Left column = the index (default = 0,1,2,3,4).

# Right column = the values from your list.

# Name: Something → the column name you gave.

# dtype: object → because of mixed types.
```

```python
#col_vec[1:4]

col_vec.loc[1:4]
```

```python
array = [1, 5, 'hello', 4, -10]
col_vec = pd.Series(array, 
                    index=['a','b','c','d','e'])
col_vec
```

```python
col_vec['c']

#slicing with step 2 - take every second element after b
#col_vec['b'::2]

#returns everything except the first row
#col_vec.iloc[1:]

#return everything starting with index c
#col_vec.loc['c':]

#col_vec.loc[['c','a']]
#col_vec.loc['c':]
```

#### Adding with `np.where()`

```python
df['new_price'] = np.where(df['A'] >= 2, 5, 3) 
df

# condition → a boolean array | 2
# value_if_true → assigned where the condition is True | 5
# value_if_false → assigned where the condition is False | 3
```

#### Adding with `.apply()`

```python
# .apply() applies a function to each element of the column 'D'
df['some_category'] = df['D'].apply(lambda x: 'Short' if x < 4 else ('Average' if x < 5 else 'Tall'))
df
```

```python
def categorize(row):
    if row['C'] < 3 and row['D'] < 5:
        return 'Light'
    elif row['C'] < 5 or row['D'] <= 6:
        return 'Middle'
    else:
        return 'Heavy'
    
df['one more category'] = df.apply(categorize, axis=1)
df
```

```python
def random_name(_):
    first_names = ['Alice', 'Bob', 'Charlie', 'David', 'Emma', 'Frank', 'Grace', 'Helen']
    last_names = ['Smith', 'Johnson', 'Williams', 'Brown', 'Jones', 'Miller', 'Davis', 'Garcia']
    return f"{np.random.choice(first_names)} {np.random.choice(last_names)}"

df['Name'] = df.apply(random_name, axis=1)

df
```

```python
df['First Name'] = df['Name'].str.split(' ').str[0]
df
```

## Grouping data

`.groupby('some_category')` groups all rows having the same attributes (as specified in by parameter) into separate data frames.

After the groupby, you need an aggregate function to summarize data in each subframe.

You may do it in 2 ways:

```python
# In each subframe, take the `price` column and summarize it
# using the `sum` function from the result
df.groupby(['some_category'])['price'].sum()
```

```python
# In each subframe, apply the `sum` function to all numeric
# columns then extract the `price` column
df.groupby(['some_category']).sum()['price']
```

```python
# Pivot Table: Think of it as a way to group your data by one or more keys and then calculate a statistic (like the mean, sum, or count)
# for each group, organized into a clean, 2D grid

df.pivot_table(index='some_category', columns='one more category', values='price', aggfunc='mean')
```
