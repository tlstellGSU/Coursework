#### 17. Create a 8x8 matrix of zeros and ones with a checkerboard pattern

```python
q17_matrix = np.zeros((8,8))
for i in range(8):
    for j in range(8):
        if (i+j) % 2 == 0:
            q17_matrix[i,j] = 1
print(q17_matrix)
```

#### 18. Create a checkerboard 8x8 matrix using the tile function

```python
q18_a = [1,0]
q18_matrix = np.tile(np.tile(q18_a, 4), 8)
q18_matrix = q18_matrix.reshape(8,8)
for i in range(8):
    if i % 2 == 1:
        q18_matrix[i] = np.flip(q18_matrix[i])
print(q18_matrix)
```

#### Check #1: Assuming identical shape of the arrays and a tolerance for the comparison of values and print result

```python
q24_A = np.random.randint(0,2,5)
q24_B = np.random.randint(0,2,5)

equal_vec = []

for i in range(len(q24_A)):
    if abs(q24_A[i] - q24_B[i]) < 1e-3:
        equal_vec.append(True)
    else:
        equal_vec.append(False)

print(all(equal_vec))
```

#### Check #2: Checking both the shape and the element values, no tolerance (values have to be exactly equal) and print result

```python
q24_A = np.random.randint(0,2,5)
q24_B = np.random.randint(0,2,5)

print(all(q24_A == q24_B))
```
