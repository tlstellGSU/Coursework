## LaTeX

### Inline Equation
You can include an inline equation by wrapping the LaTeX code with single `$` symbols.

For example: `Euler's identity: $e^{i\pi} + 1 = 0$` will display as:

Euler's identity: $e^{i\pi} + 1 = 0$

### Block Equation
To create a block equation, wrap the LaTeX code with double `$$` symbols.

For example: `$$\int_{a}^{b} f(x)dx = F(b) - F(a)$$`
will display as:
$$\int_{a}^{b} f(x) \, dx = F(b) - F(a)$$

## How to flatten an array

### `arr.reshape(-1)`

- Does NOT return a copy!

````python
arr = np.array([0, 1, 2, 3, 10, 11, 12, 13, 20, 21, 22, 23])
arr_3by4 = arr.reshape(3,4)
arr_3by4
````

````python
flat_arr = arr_3by4.reshape(-1)
print(flat_arr)
````

````python
flat_arr[0] = 100
print(flat_arr)
print(arr_3by4)
````

### `np.ravel()`

- Memory efficient
- Does NOT return a copy!

````python
arr = np.array([0, 1, 2, 3, 10, 11, 12, 13, 20, 21, 22, 23])
arr_3by4 = arr.reshape(3,4)
arr_3by4
````

````python
flat_arr = np.ravel(arr_3by4)
print(flat_arr)
````

````python
flat_arr[0] = 100
print(flat_arr)
print(arr_3by4)
````

### `arr.flatten()`

- Returns a copy

````python
arr = np.array([0, 1, 2, 3, 10, 11, 12, 13, 20, 21, 22, 23])
arr_3by4 = arr.reshape(3,4)
arr_3by4
````

````python
flat_arr = arr_3by4.flatten()
print(flat_arr)
````

````python
flat_arr[0] = 100
print(flat_arr)
print(arr_3by4)
````

## Broadcasting

Numpy arrays differ from a normal Python list because of their ability to broadcast:

````python
# Setting a value with index range (broadcasting)
arr = np.arange(0,11)
print(arr)

arr[0:5] = 100
print(arr)
````

````python
arr = np.arange(0,11)
arr[::2] = 100
print(arr)
````

````python
# Important notes on slices
arr = np.arange(0, 11)
slice_of_arr = arr[0:6]

print(slice_of_arr)
````

````python
# Change slice
slice_of_arr[:] = 99

print(slice_of_arr)
````

````python
# Now note that the changes also occur in our original array!
arr
````

Data is not copied, it's a view of the original array! This avoids memory problems!

````python
# To get a copy, need to be explicit
arr_copy = arr.copy()
arr_copy
````

## Singular Value Decomposition (SVD)

````python
A = np.array([[1, 2], [3, 4]])

# Singular Value Decomposition
U, Sigma, V = np.linalg.svd(A)

print(U)
print(Sigma)
print(V)
````

````python
Sigma_matrix = np.diag(Sigma)
Sigma_matrix
````

````python
import matplotlib.pyplot as plt
the_cat = plt.imread('./cat_image.jpg')
the_cat.shape
````

````python
U, Sigma, V = np.linalg.svd(the_cat)

reduced_images = []
sizes = [1, 2, 3, 4, 5, 10, 50, 100, 3308]
for i in sizes:
    # We choose i first rows of U
    U_reduced = U[:, :i]; 
    
    # We convert i first elements of Sigma_vector into a diagonal matrix Sigma_matrix_reduced
    Sigma_reduced = np.diag(Sigma[:i])
    
    # We choose i first columns of V
    V_reduced = V[:i, :];
    
    reduced_images.append(U_reduced @ Sigma_reduced @ V_reduced)  
````

````python
fig, ax = plt.subplots(nrows=3, ncols=3, figsize=(20,15))
i = 0
for row in ax:
    for col in row:
        current_image = reduced_images[i]
        col.imshow(current_image, cmap='gray', vmin=0, vmax=255)
        col.set_title(f"# of components = {sizes[i]}")
        col.axis('off')
        i += 1
        
# plt.savefig("reduced_image")        
````
